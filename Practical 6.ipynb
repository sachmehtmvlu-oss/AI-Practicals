Probability Reasoning (Bayesian Networks & Inference)
1. Using a simple example (Burglary–Alarm), create a Bayesian Network using
pgmpy.

```python
# Bayesian Network using pgmpy
# Example: Burglary-Alarm Network

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])

# Step 2: Define Conditional Probability Tables

# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)

# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)

# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],
        [0.001, 0.29, 0.94, 0.95]
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)

# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],
        [0.05, 0.90]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)

# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
    variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],
        [0.01, 0.70]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)

# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)

# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())

# Step 5: Perform inference
inference = VariableElimination(model)

# Query: Probability of Burglary when John and Mary both call
result = inference.query(
    variables=["Burglary"],
    evidence={
        "JohnCalls": "True",
        "MaryCalls": "True"
    }
)

print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")
print(result)
```

2. Write a program to compute Posterior Probability using Bayesian
Networks(Burglar|Alarm = True)

```python
# Posterior Probability using Bayesian Network
# Query: P(Burglary | Alarm = True)

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm")
])

# Step 2: Define probability of Burglary
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)

# Step 3: Define probability of Earthquake
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)

# Step 4: Define probability of Alarm
# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],
        [0.001, 0.29, 0.94, 0.95]
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)

# Step 5: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm
)

# Step 6: Check whether model is correct
print("Is Bayesian Network valid?", model.check_model())

# Step 7: Perform inference
inference = VariableElimination(model)

result = inference.query(
    variables=["Burglary"],
    evidence={"Alarm": "True"}
)

print("\nPosterior Probability:")
print("P(Burglary | Alarm = True)")
print(result)
```

1. Using a simple example (Rain–Sprinkler–Wet Grass), create a Bayesian Network using pgmpy.

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD

# Create the Bayesian Network
model = DiscreteBayesianNetwork([
    ("Rain", "Wet_Grass"),
    ("Sprinkler", "Wet_Grass")
])

# Probability table for Rain
cpd_rain = TabularCPD(
    variable="Rain",
    variable_card=2,
    values=[[0.8],
            [0.2]]
)

# Probability table for Sprinkler
cpd_sprinkler = TabularCPD(
    variable="Sprinkler",
    variable_card=2,
    values=[[0.6],
            [0.4]]
)

# Probability table for Wet Grass
cpd_wet_grass = TabularCPD(
    variable="Wet_Grass",
    variable_card=2,
    values=[
        [0.99, 0.10, 0.20, 0.01],
        [0.01, 0.90, 0.80, 0.99]
    ],
    evidence=["Rain", "Sprinkler"],
    evidence_card=[2, 2]
)

# Add probability tables to the model
model.add_cpds(
    cpd_rain,
    cpd_sprinkler,
    cpd_wet_grass
)

# Display the network
print("Nodes:")
print(model.nodes())

print("\nEdges:")
print(model.edges())

print("\nProbability Table for Rain:")
print(cpd_rain)

print("\nProbability Table for Sprinkler:")
print(cpd_sprinkler)

print("\nProbability Table for Wet Grass:")
print(cpd_wet_grass)

print("\nModel is valid:", model.check_model())
```

## Explanation

### Importing Libraries

- `DiscreteBayesianNetwork` is used to create the Bayesian Network.
- `TabularCPD` is used to define the probability tables.

### Creating the Network

- `Rain` and `Sprinkler` are connected to `Wet_Grass`.
- This means that both rain and the sprinkler can make the grass wet.

### Defining the Probability Tables

- `cpd_rain` defines the probability of rain.
- `cpd_sprinkler` defines the probability of the sprinkler being on or off.
- `cpd_wet_grass` defines the probability of the grass being wet based on rain and the sprinkler.

### Adding the CPDs

- `model.add_cpds()` adds all the probability tables to the network.

### Displaying the Output

- `model.nodes()` displays all the variables in the network.
- `model.edges()` displays the relationships between the variables.
- `print(cpd_rain)`, `print(cpd_sprinkler)`, and `print(cpd_wet_grass)` display the probability tables.
- `model.check_model()` verifies that the Bayesian Network has been created correctly.

### Output Explanation

- `Rain(0)` means no rain, and `Rain(1)` means rain.
- `Sprinkler(0)` means the sprinkler is off, and `Sprinkler(1)` means it is on.
- The probability tables show the likelihood of each event.
- `Model is valid: True` confirms that the network has been created successfully.

2. Write a program to compute the posterior probability using a Bayesian Network.

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create the Bayesian Network
model = DiscreteBayesianNetwork(
    [
        ("Rain", "Wet_Grass"),
        ("Sprinkler", "Wet_Grass")
    ]
)

# Probability of Rain
cpd_rain = TabularCPD(
    variable="Rain",
    variable_card=2,
    values=[[0.8], [0.2]]
)

# Probability of Sprinkler
cpd_sprinkler = TabularCPD(
    variable="Sprinkler",
    variable_card=2,
    values=[[0.6], [0.4]]
)

# Probability of Wet Grass
cpd_wet_grass = TabularCPD(
    variable="Wet_Grass",
    variable_card=2,
    values=[
        [0.99, 0.10, 0.20, 0.01],
        [0.01, 0.90, 0.80, 0.99]
    ],
    evidence=["Rain", "Sprinkler"],
    evidence_card=[2, 2]
)

# Add CPDs to the model
model.add_cpds(
    cpd_rain,
    cpd_sprinkler,
    cpd_wet_grass
)

# Check the model
print("Model is valid:", model.check_model())

# Compute posterior probability P(Rain | Wet_Grass=True)
inference = VariableElimination(model)

result = inference.query(
    variables=["Rain"],
    evidence={"Wet_Grass": 1}
)

print("\nPosterior Probability P(Rain | Wet_Grass=True)")
print(result)
```
