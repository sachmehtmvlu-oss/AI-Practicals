```python
#2. Breadth First Search (BFS) in a Maze
from collections import deque

maze = [
    [0, 0, 1],
    [1, 0, 1],
    [0, 0, 0]
]

rows = len(maze)
cols = len(maze[0])

start = (0, 0)
goal = (2, 2)

queue = deque([(start, [start])])
visited = set()

while queue:
    (x, y), path = queue.popleft()

    if (x, y) == goal:
        print("Shortest Path:", path)
        break

    if (x, y) not in visited:
        visited.add((x, y))

        for dx, dy in [(0,1),(1,0),(0,-1),(-1,0)]:
            nx = x + dx
            ny = y + dy

            if 0 <= nx < rows and 0 <= ny < cols:
                if maze[nx][ny] == 0 and (nx, ny) not in visited:
                    queue.append(((nx, ny), path + [(nx, ny)]))
```

```python
#4. Missionaries and Cannibals using BFS
from collections import deque

start = (3, 3, 1)
goal = (0, 0, 0)

moves = [(1,0),(2,0),(0,1),(0,2),(1,1)]

def valid(state):
    m, c, b = state

    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    if m > 0 and m < c:
        return False

    if (3-m) > 0 and (3-m) < (3-c):
        return False

    return True

queue = deque([(start, [start])])
visited = set()

while queue:
    state, path = queue.popleft()

    if state == goal:
        print("Solution:")
        for s in path:
            print(s)
        break

    if state not in visited:
        visited.add(state)

        m, c, b = state

        for dm, dc in moves:
            if b == 1:
                new = (m-dm, c-dc, 0)
            else:
                new = (m+dm, c+dc, 1)

            if valid(new):
                queue.append((new, path+[new]))
```

```python
#5. A* Search and Comparison with UCS
import heapq

graph = {
    'A': [('B',1), ('C',4)],
    'B': [('D',2), ('E',5)],
    'C': [('F',3)],
    'D': [],
    'E': [('F',1)],
    'F': []
}

heuristic = {
    'A':7,
    'B':6,
    'C':2,
    'D':1,
    'E':1,
    'F':0
}

def astar(start, goal):
    pq = [(0, start)]
    cost = {start:0}

    while pq:
        f, node = heapq.heappop(pq)

        if node == goal:
            print("Goal Reached")
            print("Cost =", cost[node])
            return

        for nxt, w in graph[node]:
            new_cost = cost[node] + w

            if nxt not in cost or new_cost < cost[nxt]:
                cost[nxt] = new_cost
                priority = new_cost + heuristic[nxt]
                heapq.heappush(pq, (priority, nxt))

astar('A', 'F')
```
