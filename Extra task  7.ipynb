```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create DataFrame
data = {
    "Area": [600, 800, 1000, 1200, 1400, 1600],
    "Price": [32, 40, 49, 58, 68, 79]
}

df = pd.DataFrame(data)

# Independent and Dependent variables
X = df[['Area']]
y = df['Price']

# Train model
model = LinearRegression()
model.fit(X, y)

# Regression Equation
m = model.coef_[0]
c = model.intercept_

print("Regression Equation:")
print(f"Price = {m:.4f} * Area + {c:.4f}")

# Prediction
price1500 = model.predict([[1500]])
price2500 = model.predict([[2500]])

print("\nPredicted Price for 1500 sq.ft.:", round(price1500[0],2), "Lakhs")
print("Predicted Price for 2500 sq.ft.:", round(price2500[0],2), "Lakhs")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# 3. Load the Employee Salary Dataset
data = pd.read_csv("employee_salary_dataset.csv")

# Display first five records
print("First Five Records:")
print(data.head())

# Check missing values
print("\nMissing Values:")
print(data.isnull().sum())

# 4. Select Independent (X) and Dependent (Y) Variables
X = data[['Experience_Years']]
y = data['Monthly_Salary']

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)

# 5. Train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# 6. Predict salary for an employee with 7 years of experience
new_data = pd.DataFrame([[7]], columns=['Experience_Years'])
predicted_Monthly_Salary = model.predict(new_data)

print("\nPredicted Salary for 7 Years Experience:", round(predicted_Monthly_Salary[0], 2))

# 7. Display Coefficient, Intercept and R² Score
print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

y_pred = model.predict(X_test)
print("R² Score:", round(r2_score(y_test, y_pred), 2))

# 8. Plot the Regression Line
plt.figure(figsize=(8,6))
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, model.predict(X), color='red', label='Regression Line')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Employee Salary Prediction using Linear Regression")
plt.legend()
plt.show()
```
